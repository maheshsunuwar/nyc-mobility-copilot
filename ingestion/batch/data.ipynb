```python
import pyarrow.parquet as pq

y_schema = pq.read_table('yellow_tripdata_2026-01.parquet').schema
g_schema = pq.read_table('green_tripdata_2026-01.parquet').schema

y_cols = set(y_schema.names)
g_cols = set(g_schema.names)

print('Yellow Columns: ', y_schema.names)
print()
print('Green Columns: ', g_schema.names)
print()
print('In yellow but not green:', y_cols - g_cols)
print('In green but not yellow:', g_cols - y_cols)
```
